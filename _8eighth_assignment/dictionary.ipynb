1. what type of data structure is a dictionary in python ?
- A dictionary is a set of key , value pairs to be separated with colon.

  the keys are immutable and used as hash values to efficiently access the values with lesser 
  time complexity. keys cannot be duplicate 

  the values are mutable and can be duplicate 

  we can iterate over a dictionary using for loop and access the values by providing key in the 
  subscripting brackets.


2. Are  dictionaries mutable or immutable ?
- Dictionaries are mutable in nature that we can add, remove and update any element present in  
  it.
  
  the keys are immutable and cannot be of any mutable type such as lists or sets. It can be tuple or frozenset or string that are simply hashable.

  the values are mutable that means it can be either lists or sets and can be changed in the mean time code is running.

3. What separator is used between key and value in a dictionary ?
-  The colon separator is used when we create dictionary using paranthesis and equals to 
   sign while using the dict() constructor.

4. Can dictionary keys be duplicated ?
- The dictionary keys cannot be duplicate since they are immutable and are hashed to operate 
  providing a duplicate value overwrites the previous one and the most recent one is considered.

5. Can dictionary values be duplicated ?
- The dictionary values can be duplicate values since they are mutable and not required to be  
  hashed they can store duplicate values but keys must be unique.

6. Are dictionaries ordered or unordered conceptually ?
- The dictionaries are ordered and preserves the order of the insertion of pairs but we cannot 
  access the values or items using positional number such as indexing which we use in list.

7. Which functions is used to safely access a key without raising an error ?
-  we use .get() method to safely access a key.

8. Which dictionary method removes the last inserted item ?
-  The .popitems() is used to remove the last inserted item from a dictionary.

9. Can a list be used as a dictionary key ?
- No we cannot use list or set as a key in the dictionary because both are mutable and 
  unhashable.

10. Can a frozenset be used as a dictionary key?
- Yes we can use a frozenset as a key in dicionary because it is immutable and hashable.
