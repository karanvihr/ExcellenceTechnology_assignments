```python
#  Area of either square or rectangle
shape = input("enter the shape to find area from the given shapes :" \
"1. square \n 2. rectangle \n 3. circle \n 4. triangle \n ")

def square():
    num = int(input("enter the side of the square"))

    return num*num,num*4

def rectangle():
    l = int(input("enter the length of the rectangle"))
    b = int(input("enter the breadth of the rectangle"))
    return l*b,2*(l*b) 

def circle():
    r =int(input("enter the radius of the circle"))
    return (22/7)*r*r,2*(22/7)*r

def triangle():
    base = int(input("enter the base of the triangle"))
    height = int(input("enter the height of the triangle"))
    s1 = int(input("enter the side 1 "))
    s2 = int(input("enter the side 2 "))
    s3 = int(input("enter the side 3 "))
    return (base*height/2,s1+s2+s3)

match shape:
    case "square":
       area,per =  square()
    case "rectangle":
       area,per =  rectangle()
    case "triangle":
       area,per = triangle()
    case "circle":
       area,per =  circle()            
    
print("the area of the ",shape," is ",area,"& perimeter is ",per)



```
