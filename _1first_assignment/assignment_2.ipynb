```python
# 2. Assignment Operators

#  Use += operator      
num = 10
num += 10
print(num)

# Use -= operator
num = 10
num -= 10
print(num)

# Use *= operator
num = 10
num *= 10
print(num)

# Use /= operator
num = 10
num /= 10
print(num)

```

```python
# Create a running Total
total = int(input("enter a number"))

for i in range(4):
 oper = input("enter a operation to perform over number")
 num = int(input("enter a number to run"))


 if oper == '+':
     total += num
     print("total: ",total)

 if oper == '-':
     total -= num
     print("total: ",total)

 if oper == '*':
     total *= num
     print("total: ",total)

 if oper == '/':
     total /= num  
     print("total: ",total)
```

```python
# Update salary system
salary = int(input("enter the salary"))
oper = (input("enter\n '+' to increment \n '-' to decrement "))
num = int(input("enter the value to be updated"))

if oper == '+':
    salary += num
elif oper == '-':
    salary -= num

print("the Updated Salary is ",salary)
```

```python
# Inventory update system
inventory = int(input("enter the no. of items in inventory"))
oper = (input("enter\n '+' to increment \n '-' to decrement "))
num = int(input("enter the value to be updated"))

if oper == '+':
    inventory += num
elif oper == '-':
    inventory -= num

print("the Updated No. of Items in  Inventory is ",inventory)
```

```python
# Attendance Counter system
atten = int(input("Enter the current Attendance"))

while True:
    pre = input("enter P -present or A -absent \n Enter 'exit' to stop the attendance counter")
    if pre =='P':
        atten+=1
    if pre =='A':
        pass 
    if pre =='exit':
        break
    print("Attendance : ",atten)
```
