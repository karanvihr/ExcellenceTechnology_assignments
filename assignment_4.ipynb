```python
# 4. Logical Operators

num1 = 100 
num2 = 200
# Use AND operator
if num1>500 and num2>500 :
    print("both the numbers are greater than 500")
# Use OR operator 
if num1<500 or num2<500 :
    print("either of the number is  less than 500")
# Use NOT operator
if not num1>500 and not num2>500 :
    print("both the numbers are smaller than 500")
```
