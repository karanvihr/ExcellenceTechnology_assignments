```python
# coupon validation
coupons ={
    "first":[1912,9812,9012,4312],
    "rakhi":[1244,3244,8844,1844],
    "special":[3025,1425,5425,9825],
    "guest":[2121,4321,4521,5621]
}

# validate coupon entered by user from the above given dicitionery
usr_coupon = int(input("enter your coupon code"))

for key in coupons:
 if usr_coupon in coupons[key]:
  print("this is a valid coupon",usr_coupon,"from:",key)
```

```python
# answer checking using membership

quiz = {
    "capital of india":["delhi","new delhi"],
    "largest planet of solar system":["jupiter"],
    "name of python founder":["guido van rossum","guido"]
}

for key in quiz:
    ans = input(f"what is the {key}")
    if ans in quiz[key]:
        print("answer is correct.")
    else:
        print("incorrect answer")   
```

```python
# hashtag detection from the data string
# i love #python and #coding
text = input("enter the text")

words = text.split()
# now make a list of words having #
hashtags = [word for word in words if "#" in word] 

print("Hashtags found are ",hashtags)
```

```python
# Spam detection 

spams = ["free","win","offer","click","money","prize"]

user_msg = input("enter a message").lower()

if any(keyword in user_msg for keyword in spams):
    print("the user is spaming")
else:
    print("message is safe")
```
