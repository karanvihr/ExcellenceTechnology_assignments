```python
# find the first occurence of a substring
str20 = "findthefirstoccurence"
print(str20.find("x"))

# find the position of a substring 
print(str20.index("occ"))

# find the last occurence of a substring
print(str20.rfind("f"))

# count occurences of a character
print(str20.count("f"))

# replace a word in a string
print(str20.replace("f","g"))
```

```python
# remove leading and trailing spaces using strip()
str21 = "    htos is free space   "
print(str21.strip())


# remove only right-side spaces using
print(str21.rstrip())

# remove only left-side spaces using
print(str21.lstrip())
```

```python
# check whether a string starts with a specific word 

str22 = "Helloworld"
print(str22.startswith("hell"))

# check whether a string ends with a specific word 
print(str22.endswith("rld"))


# check whehter a string contains only alphabets
print(str22.isalpha())

# check wheter a string contains only digits
print(str22.isdigit())

# check whether a string is alphanumeric
print(str22.isalnum())

# check whether a string is title
print(str22.istitle())

# check whether a string is uppercase
print(str22.isupper())

# check whether a string is lowercase
print(str22.islower())
```
