```python
#Q. Define a Set in Python. Write any four characteristics of sets
#Ans.  A set is an unordered , mutable list of heterogenous, unique elements 
#      it must contains hashable elements such as integers , tuples , strings
#      it is defined using curly braces and must contain one element . empty curly 
#      braces are considered as dictionary 

# Q. Create a set with duplicate values and show how duplicates are removed automatically.
# Ans. The set  does not contain duplicate values so it reomves the duplicate values provided
mySet = {1,2,3,1,3,4,5,6} 
print(mySet)
#       A set stores its elements after hashing them to create a hash value that automaticaly
#       removes values of similar hash value. thus only create a single occurence of an element
#       throughout the set
```

```python
# Q. Differentiate between List and Set with Examples

myList = [1,2,3,4,5,5,5,1]
# a list can store multiple same values and and can have elements index value to fetch it.
mySet = {11,22,33,44,55,55}
# a set can only store single occurence and connot access elements using index values.

myList.index(5)
# mySet.index() such method exists 

mySet.add(10)
myList.append(10)
print(myList)
print(mySet)

# in list we can change elements 
myList[0] = 111
# in set we can only add or remove an element but cannot change an element
# mySet[0] = 111
#               - indexing of an element in sets is not allowed
```

```python
# Q. Why is Indexing and Slicing not allowed in sets?
# Ans. The sets are unordered list of hetrogenous and unique element
#      also these elements are stored in hash values making them directly inaccessible and usable
mySet = {"1","2","3b"}
print(mySet)


# Q. Write a program to add an element to set
mySet.add(21)
print(mySet)

# Q. Write a program to add multiple elements to a set using update
sample = [101,102,103,104]
mySet.update((111,112,113,114))
mySet.update(sample)
print(mySet)


# Q. Explain the difference between remove() and discard()
mySet.remove(101) #this removes a member of the set if exists or raise an error
mySet.discard(111) #this also removes a member of the set but unlike remove it does not raise an error
print(mySet)

# Q. Write a program to remove a random element from a set 
mySet.pop() #this removes a random element from the set 

# Q. Write a program to create a copy of a set and compare the original and copied sets
mySet2 = mySet.copy()  # this returns a shallow copy of the set
mySet3 = mySet
print(mySet == mySet2)    #contents are the same (True)
print(mySet is mySet2)    #different objects in the memory (False)

print(mySet == mySet3)    #contents are the same
print(mySet is mySet3)    #same object in memory
```

```python
# Q. Write a program to find the union of two sets
mySet = {1,2,3,4,5,6,7,8,9}
mySet2 = {6,7,8,9,10,11,12}

print(mySet.union(mySet2))


# Q. Write a program to find the intersection of two sets
print(mySet.intersection(mySet2))


# Q. Write a program to find the difference and symmetric difference of two sets
print(mySet.difference(mySet2))
print(mySet.symmetric_difference(mySet2))


# Q. Write a program to remove duplicate values from a list using set
myList = [1,2,1,3,4,3,4,5,6,7,7]
mySet= set(myList)
myList = mySet
print(myList)
```
