```python
# 6. Read using readline() to read and display them.
# create a file containing at least 5 lines

with open('abc.txt','w')as file:
    file.write("lorem ipsum de set color.1\n")
    file.write("lorem ipsum de set color.2\n")
    file.write("lorem ipsum de set color.3\n")
    file.write("lorem ipsum de set color.4\n")
    file.write("lorem ipsum de set color.5\n")
    print("success")

with open('abc.txt','r') as file:
    print(file.readline(26))
    print(file.readline(26))
```

```python
# 7. Read multiple lines using readline()/
# use readline() to read the first three lines of a file 

with open("abc.txt",'r') as file :
    print(file.readline(26))
    print(file.readline(26))
    print(file.readline())
```

```python
# 8. Read Using  readlines()
# create a file containing atleast 5 lines.
with open('abc.txt','r')as file:
    data = file.readlines()

for line in data:
    print(line)
```

```python
# 9. Difference between readline and readlines 
with open('abc.txt','r') as file:
    print(file.readline())
    data = file.readlines()

for l in data:
    print(l)


"""the readline() method returns only single line from the file and keeps the teel pointer on the next line so that every time we rerun the readline()
method it fetchs a new next line in a sequence and when the tell pointer is at the last of file EOF it returns nothing.
whereas readlines() method returns a list of items each item is a line an individual line from the file separated and indexed.
the difference in both funtions is one works with sequenced data accessing other access all the data at once and then iterating over it using indexes."""
```

# Section C - Append

```python
# 10. Append Data
# create a file named abc.txt and write the details of one student.
# then use append to add details of two more students

with open('abc.txt','w')as file:
    file.write("Name : karan\tage: 22\tcourse: python\n")
    print("successfuly added a student.")

with open('abc.txt','r') as file:
    data = file.read()
    print(data)

with open('abc.txt','a') as file:
    file.write("Name: rahul\tage: 21\tcourse: python\n")
    file.write("Name: rakul\tage: 22\tcourse: photon\n")
    file.write("Name: basul\tage: 11\tcourse: pyton\n")
    print("successfully added.")

with open('abc.txt','r') as file:
    data = file.read()
    print(data)
```

```python
# 11. add new information
# create new file write some note into it then use apend to add three more notes.
with open('abc.txt','w')as file :
    file.write("today is a lucky day.\n")

with open('abc.txt','a')as file:
    file.write('also tommorow will be also good day.\n')
    file.write('also day after tommorow will be also good day.\n')
    file.write('also day after day after tommorow will be also good day.\n')

with open('abc.txt','r')as file:
    data = file.read()
    print(data)
```
