```python
# 6. Calculator Create a calculator that performs: text .Addition .Subtraction .Multiplication .Division

class calculator:

    def addition(x,y):
        return x+y

    def subtraction(x,y):
        return x-y

    def multiplication(x,y):
        return x*y

    def division(x,y):
        return x//y

    def userinput():
        ans = 0
        try:
            n1 = int(input("enter a number."))
            op = input("enter operation. ' + '   ' - '   ' * '   ' / '")
            n2 = int(input("enter a number."))
            
            # remove any white spaces or capitals 
            # op = str(op).islower()
            if op[0] == " " or op[-1] == " ":
                op = op.strip()
            
            if op == '+' or op =='add' or op == 'addition' or op == 'plus':
                ans = calculator.addition(n1,n2)

            if op == '-' or op == 'sub' or op == 'subtract' or op == 'subtraction' or op == 'minus':
                ans = calculator.subtraction(n1,n2)

            if op == '*' or op == 'mul' or op == 'multiply' or op == 'multiplication' or op == 'into':
                ans = calculator.multiplication(n1,n2)

            if op == '/' or op == '//' or op == 'div' or op == 'divide' or op == 'division':
                try:
                    ans = calculator.division(n1,n2) 
                except Exception as e:
                    print('karan')
            return ans
        
        except ValueError as e:
            print(e)

val = calculator.userinput()
if val is not None:
    print(val)  
```

```python
# 17. create  a custom exception 
# invalidageerror

class InvalidAgeError(Exception):
    pass

try:
    age = int(input("enter age"))
    if age < 0 or age > 100:
        raise InvalidAgeError("age connot be zero or more than hundred.")
    print("your age:",age)
except InvalidAgeError as e:
    print(e)
    
```

```python
# 19. ATM Program 

class BankOfEt:

    __balance = 10000

    @classmethod
    def deposit(cls,cash):
        try:
            if cash <0:
                raise ValueError("cannot withdraw cash in negative.")
            cls.__balance += cash

            
            return True,f"balance is {cls.__balance}"
        except ValueError as e:
            return False,f'{e}'


    @classmethod
    def withdraw(cls,cash):
        try:
            if balance > cash:
                if cash < 0:
                    raise ValueError("cannot withdraw negative cash.")

                cls.__balance -=cash
                return True,f"withdrawn Succesful. balance: {cls.__balance}"

                
            else:
                return False,f"Insufficient Balance."
            

        except :
                return False,f"Cannot withdraw negative."

    @classmethod
    def checkbalance(cls):
        return True,f"your balance is {cls.__balance}"
    
    def atm():

        exit = True
        while exit:

            put = input("enter\n1. Deposit \n2. Withdraw \n3.Chech Balance \n4.Exit")

            if put == '1':
                cash = int(input("enter amount to deposit."))
                success,message = BankOfEt.deposit(cash=cash)
                if success:
                    print(message)
                else:
                    print(message)

            if put == '2':
                cash = int(input("enter amount to withdraw."))
                success,message = BankOfEt.withdraw(cash=cash)
                if success:
                    print(message)
                else:
                    print(message)

            if put == '3':
                success,message = BankOfEt.checkbalance()
                if success:
                    print(message)
            if put == '4':
                exit = False




BankOfEt.atm()
```

```python
# 20. Student Marks program
try: 
    name = input('enter your name.')
    marks = []
    
    while len(marks) < 5:
        marks.append(int(input("enter marks in a subject.")))

    for i in marks:
        if i < 0 or i > 100:
            raise ValueError("Marks cannot be less zero or greater than 100.")

    print(f"percentage of {name} : {sum(marks)/5}")
except Exception as e:
    print(name,"",e)
    
```
