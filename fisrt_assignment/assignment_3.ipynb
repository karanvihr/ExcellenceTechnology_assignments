```python
# Result Validation 

m1 = int(input("enter your marks in subject:1"))
m2 = int(input("enter your marks in subject:2"))
m3 = int(input("enter your marks in subject:3"))
m4 = int(input("enter your marks in subject:4"))
m5 = int(input("enter your marks in subject:5"))

total = m1+m2+m3+m4+m5

result = (total/500)*100
result = int(result)

if result>90:
    print(result,"you have scored A+ in the exam")

elif result>80:
    print(result,"you have scored A in the exam")

elif result>65:
    print(result,"you have scored B+ in the exam")

elif result>33:
    print(result,"you have passed the exam")        
```
