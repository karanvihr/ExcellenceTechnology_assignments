```python
# check whether a year  is leap year or not
year = int(input("enter a year"))
# print("to check a year is leap year it should be divisible by 4 and not divisible by 100 and if it should be divisble by 400.")
if (year%4==0 and year%100!=0)or(year%400==0):
    print("leap year")
else:
    print("not leap year")
```
