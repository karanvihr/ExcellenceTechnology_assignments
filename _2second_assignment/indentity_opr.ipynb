```python
# 7. Identity Operators 
# Compare objects using is operator
user = {
    "fisrt_name":"karan",
    "middle_name":"veer",
    "last_name":"singh"
}

user1 = {"first_name":user["fisrt_name"],
         "middle_name":user["middle_name"],
         "last_name":user["last_name"]}

user2 = user

if user2 is user:
 print(True)
else :print(False)

if user1 is user:
 print(True)
else :print(False)


if user1 is user2:
 print(True)  
else :print(False)

```

```python
# Compare objects using is not
user = {
    "fisrt_name":"karan",
    "middle_name":"veer",
    "last_name":"singh"
}

user1 = {"first_name":user["fisrt_name"],
         "middle_name":user["middle_name"],
         "last_name":user["last_name"]}

user2 = user

print(True)if user2 is not user else print(False) 
print(True)if user1 is not user else print(False) 
print(True)if user1 is not user2 else print(False) 
```

```python
# Singleton example 

class singleton:
    _instance = None #class levvel variable to store single instance

    def __new__(cls,*args,**kwargs):
        if cls._instance is None:  #if no instance exists
            cls._instance = super().__new__(cls)  #create one

        return cls._instance #always return the same instance

obj1 = singleton()
obj2 = singleton()

print(True)if obj1 is obj2 else print(False)
```

```python
# Configuration manager instance
class ConfigManager:
    _instance = None
    def __new__(cls,*args, **kwargs):
        if cls._instance is None:
            print("Loading configuration into memory...")
            cls._instance = super().__new__(cls)
        return cls._instance

c1 = ConfigManager()
c2 = ConfigManager()

print(c1 is c2)

```

```python
# Aliasing vs Copy
x = [10,20,30]
y = x
z = x.copy()

y.append(40)
z.append(50)
z.append(60)

print("x: ",x)
print("y: ",y)
print("z: ",z)
```

```python
# Stages of Object Lifecycle

# 1. Creation
# Objects are created when you instantiate class
class Student:
    def __init__(self,name):
        self.name = name
        print(f"student {self.name} created")
    def __del__(self):
        print(f"Student {self.name} destroyed")

s1 = Student("Karanveer")

# 2. Usage
# Once created, the object can be used 
# -accessing attributes, calling methods, etc.
print(s1.name)  #usage

# 3. Destruction
# When an object is no longer referenced, Python's 
# garbage collector destroys it. you can observe this with 
# the __del__ method.
s2 = Student("Aliasing Example")
del s2   #Explicitly delete reference

```
