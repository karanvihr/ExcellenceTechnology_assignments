```python
# Binary Calculator
# taking user input
o = input("enter the Bitwise Operator").strip().upper()

# mapping operation names ("AND" "OR" "NOT" "XOR" "LEFT_SHIFT" "RIGHT_SHIFT")

# define bitwise operations
bitwise_oprs = {
    "AND": lambda x,y:x&y,
    "OR":  lambda x,y:x|y,
    "NOT": lambda x: ~x,
    "XOR": lambda x,y:x^y,
    "LFT_SHFT": lambda a,n : a<<n,
    "RHT_SHFT":lambda a,n : a>>n,
}

# Handling operations based on the input requirements
if o == "NOT":
    a = int(input("enter a number"))
    result = bitwise_oprs[o](a)
else:
    if o in ["LFT_SHFT","RHT_SHFT"]:
        a = int(input("enter a number"))
        n = int(input("enter shift amount"))
        result = bitwise_oprs[o](a,n)
    else:
        x = int(input("enter a number"))
        y = int(input("enter b number"))
        result = bitwise_oprs[o](x,y)    

print(result)
print("Binary: ",bin(result))


```

```python
# Defining Global Permissions constants
READ    = 4
WRITE   = 2
EXECUTE = 1

# Defining a class permissions
class permissions:
    def __init__(self,perms=0):
        self.perms = perms
    def add(self,perms):
        self.perms |= perms    #set Bit
    def remove(self,perms):
        self.perms &= ~perms   #clear Bit
    def has(self,perms):
        return bool(self.perms & perms) #check Bit
    def __str__(self):
        return f"permissions: {bin(self.perms)}"


userPerms = permissions()

# ADD Read and Write permisions
userPerms.add(READ)
userPerms.add(WRITE)
print(userPerms)

# check if user has write permission
print("Write allowed: ",userPerms.has(WRITE))
print(userPerms)
# remove write permission
userPerms.remove(WRITE)
print(userPerms)
```
