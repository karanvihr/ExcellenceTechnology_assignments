```python
# create a new fruit to the list using append()
fruits = list()
fruits.extend(["Apple","Mango","Banana","Guava","Peache"])
print("before adding the fruit: 'Pomegranate'")
for fruit in fruits:
    print(fruit)
print()
print("adding pomegranate...")
print()
fruits.append("Pomegranate")
for fruit in fruits:
    print(fruit)

# Remove an element from the list
fruits.remove("Apple")
print("Removing apple...")
print()
for fruit in fruits:
    print(fruit)
```

```python
# write a program to find duplicate elements in a list 
list4 = [1,2,6,3,4,5,6,6,6,7,8,10,1]
list4.sort()
c =0
for i in range(len(list4)):
    if list4[i-1] == list4[i]:
     c +=1
     print(f"the number {list4[i]} is repeated")
```

```python
# merge two lists into one
list7 = [1,2,3,4,5,6]
list8 = [7,8,9]
list7.extend(list8)
print(list7)


# remove all even numbers in the list
for n in list7:
    if n%2==0:
        list7.remove(n)

print(list7)
```
